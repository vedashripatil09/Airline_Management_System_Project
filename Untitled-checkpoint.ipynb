```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mysql.connector
```

```python
query = "SELECT * FROM bookings"

bookings = pd.read_sql(query, conn)

bookings.head()
```

```python
query = "SELECT * FROM flights"

flights = pd.read_sql(query, conn)

flights.head()
```

```python
query = "SELECT * FROM passengers"

passengers = pd.read_sql(query, conn)

passengers.head()
```

```python
query = """
SELECT p.name, f.airline, f.source_city, f.destination_city, b.ticket_price
FROM bookings b
JOIN passengers p ON b.passenger_id = p.passenger_id
JOIN flights f ON b.flight_id = f.flight_id
"""

data = pd.read_sql(query, conn)

data.head()
```

```python
revenue = data.groupby("airline")["ticket_price"].sum().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=revenue, x="airline", y="ticket_price")

plt.title("Revenue by Airline")
plt.xlabel("Airline")
plt.ylabel("Total Revenue")

plt.show()
```

```python
routes = data.groupby(["source_city","destination_city"]).size().reset_index(name="total_bookings")

routes = routes.sort_values("total_bookings", ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(data=routes, x="total_bookings", y="source_city")

plt.title("Top Travel Routes")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data["ticket_price"], bins=20)

plt.title("Ticket Price Distribution")

plt.show()
```

```python
bookings_by_airline = data["airline"].value_counts()

plt.figure(figsize=(8,5))
bookings_by_airline.plot(kind="bar")

plt.title("Bookings per Airline")

plt.show()
```
